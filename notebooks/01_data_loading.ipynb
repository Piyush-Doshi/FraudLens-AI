```python
import pandas as pd
import os
```

```python
accounts_path = r"D:\Projects\major project\FraudLens-AI\Data\Raw_Data\HI-Small_accounts.csv"
transactions_path = r"D:\Projects\major project\FraudLens-AI\Data\Raw_Data\HI-Small_Trans_Custom.csv"
```

```python
accounts = pd.read_csv(accounts_path)
transactions = pd.read_csv(transactions_path)
```

```python
print("Accounts dataset loaded:", accounts.shape)
print("Transactions dataset loaded:", transactions.shape)
```

```python
print("===== DATASET SIZE =====")

print(f"Accounts Rows      : {accounts.shape[0]:,}")
print(f"Accounts Columns   : {accounts.shape[1]:,}")

print(f"Transaction Rows   : {transactions.shape[0]:,}")
print(f"Transaction Columns: {transactions.shape[1]:,}")
```

```python
def get_file_size(file_path):
    size_bytes = os.path.getsize(file_path)
    
    if size_bytes < 1024**2:
        return f"{size_bytes / 1024:.2f} KB"
    elif size_bytes < 1024**3:
        return f"{size_bytes / 1024**2:.2f} MB"
    else:
        return f"{size_bytes / 1024**3:.2f} GB"


print("===== FILE SIZE =====")

print("Accounts dataset    :", get_file_size(accounts_path))
print("Transactions dataset:", get_file_size(transactions_path))
```

```python
print("===== ACCOUNTS ATTRIBUTES =====")

for i, column in enumerate(accounts.columns, start=1):
    print(f"{i}. {column}")
```

```python
accounts_summary = pd.DataFrame({
    "Attribute": accounts.columns,
    "Data Type": accounts.dtypes.astype(str).values,
    "Non-Null Count": accounts.notna().sum().values,
    "Missing Count": accounts.isna().sum().values,
    "Unique Values": accounts.nunique().values
})

accounts_summary
```

```python
transactions_summary = pd.DataFrame({
    "Attribute": transactions.columns,
    "Data Type": transactions.dtypes.astype(str).values,
    "Non-Null Count": transactions.notna().sum().values,
    "Missing Count": transactions.isna().sum().values,
    "Unique Values": transactions.nunique().values
})

transactions_summary
```

```python
print("===== ACCOUNTS DATA TYPES =====")
print(accounts.dtypes)

print("\n===== TRANSACTIONS DATA TYPES =====")
print(transactions.dtypes)
```

```python
accounts_missing = pd.DataFrame({
    "Attribute": accounts.columns,
    "Missing Count": accounts.isna().sum().values,
    "Missing Percentage": (
        accounts.isna().mean().values * 100
    ).round(2)
})

accounts_missing
```

```python
transactions_missing = pd.DataFrame({
    "Attribute": transactions.columns,
    "Missing Count": transactions.isna().sum().values,
    "Missing Percentage": (
        transactions.isna().mean().values * 100
    ).round(2)
})

transactions_missing
```

```python
account_duplicates = accounts.duplicated().sum()

print("Duplicate account records:", account_duplicates)
```

```python
transaction_duplicates = transactions.duplicated().sum()

print("Duplicate transaction records:", transaction_duplicates)
```

```python
print(transactions.columns.tolist())
print(accounts.columns.tolist())
```

```python
print("Unique Sender Accounts:",
      transactions["Account"].nunique())

print("Unique Receiver Accounts:",
      transactions["Account.1"].nunique())
```

```python
sender_accounts = set(transactions["Account"].dropna().unique())
receiver_accounts = set(transactions["Account.1"].dropna().unique())

transaction_accounts = sender_accounts.union(receiver_accounts)

print("Total Unique Accounts in Transactions:",
      len(transaction_accounts))
```

```python
for column in accounts.columns:
    print(
        column,
        "→",
        accounts[column].nunique(),
        "unique values"
    )
```

```python
account_master_set = set(
    accounts["Account Number"].dropna().unique()
)

print("Unique accounts in Accounts dataset:",
      len(account_master_set))
```

```python
sender_not_found = (
    sender_accounts - account_master_set
)

receiver_not_found = (
    receiver_accounts - account_master_set
)

print("Sender accounts not found:",
      len(sender_not_found))

print("Receiver accounts not found:",
      len(receiver_not_found))
```

```python
print("===== GRAPH CONNECTIVITY =====")

print("Unique Sender Accounts  :", len(sender_accounts))
print("Unique Receiver Accounts:", len(receiver_accounts))
print("Total Unique Accounts   :", len(transaction_accounts))
print("Total Transactions      :", len(transactions))
```

```python
part_a_summary = pd.DataFrame({
    "Metric": [
        "Accounts Rows",
        "Accounts Columns",
        "Transaction Rows",
        "Transaction Columns",
        "Account Duplicates",
        "Transaction Duplicates",
        "Unique Transaction Accounts"
    ],
    "Value": [
        accounts.shape[0],
        accounts.shape[1],
        transactions.shape[0],
        transactions.shape[1],
        accounts.duplicated().sum(),
        transactions.duplicated().sum(),
        len(transaction_accounts)
    ]
})

part_a_summary
```

## Part A — Dataset Understanding Summary

### Datasets Used

1. `HI-Small_accounts.csv`
2. `HI-Small_Trans_Custom.csv`

### Findings

- The accounts dataset provides account-level information.
- The transaction dataset contains relationships between sender and receiver accounts.
- The transaction data can be represented as a directed graph.
- Accounts will be represented as graph nodes.
- Transactions will be represented as graph edges.
- The original transaction fields `Account` and `Account.1` will later be represented in the database as `Sender_Account` and `Receiver_Account`.
- The exact relationship between the two datasets must be validated before database creation.
